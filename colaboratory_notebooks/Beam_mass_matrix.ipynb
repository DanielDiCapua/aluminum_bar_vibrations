```python
from sympy import *
import math
```

```python
EA, A, L, EI, rho, c, s = symbols('EA, A, L, EI, rho, c, s')
```

```python
K_la = zeros(6,6)
K_la[0,0] =  EA/L
K_la[0,3] = - EA/L
K_la[3,0] = -EA/L
K_la[3,3] = EA/L
```

```python
fact = EI/(L**3)
K_lf = zeros(6,6)
K_lf[1,1] = fact*12
K_lf[1,2]= fact*6*L
K_lf[1,4]= -fact*12
K_lf[1,5]= fact*6*L

K_lf[2,1] = fact*6*L
K_lf[2,2]= fact*4*L**2
K_lf[2,4]= -fact*6*L
K_lf[2,5]= fact*2*L**2

K_lf[4,1] = -fact*12
K_lf[4,2]= -fact*6*L
K_lf[4,4]= fact*12
K_lf[4,5]= -fact*6*L

K_lf[5,1] = fact*6*L
K_lf[5,2]= fact*2*L**2
K_lf[5,4]= -fact*6*L
K_lf[5,5]= fact*4*L**2

```

```python
Kl = K_la + K_lf
```

```python
T = eye(6)
T[0,0]= c
T[0,1]= s
T[1,0]= -s
T[1,1]= c
T[3,3]= c
T[3,4]= s
T[4,3]= -s
T[4,4]= c
T
```

```python
Kg = T.T*Kl*T
Kg
```

```python
Kg.subs({EA:80000000, A: 0.02, L: 3, EI: 640000, rho:2500, c: 0.707106781, s: 0.707106781}).evalf()
```

```python
M_la = zeros(6,6)
factMla = rho*A*L/6
M_la[0,0] = 2*factMla
M_la[0,3] = factMla
M_la[3,0] = factMla
M_la[3,3] = 2*factMla
```

```python
M_lf = zeros(6,6)
factMlf = rho*A*L/420
M_lf[1,1] = factMlf * 156
M_lf[1,2] = factMlf * 22 * L
M_lf[1,4] = factMlf * 54
M_lf[1,5] = -factMlf * 13 * L

M_lf[2,1] = factMlf * 22 * L
M_lf[2,2] = factMlf * 4 * L**2
M_lf[2,4] = factMlf * 13 * L
M_lf[2,5] = -factMlf * 3 * L**2

M_lf[4,1] = factMlf * 54
M_lf[4,2] = factMlf * 13 * L
M_lf[4,4] = factMlf * 156
M_lf[4,5] = -factMlf * 22 * L

M_lf[5,1] = -factMlf * 13 * L
M_lf[5,2] = -factMlf * 3 * L**2
M_lf[5,4] = -factMlf * 22 * L
M_lf[5,5] = factMlf * 4 * L**2
```

```python
M = M_la + M_lf
Mg = T.T*M*T
Mg
```
